## P1_0 ==> P(px=1|0)
## P0_0==> p(px=0|0) =1-P(px=1|0)
## P1_1 ==>P(px=1|1)
## P1_0==> p(px=0|1) =1-P(px=1|1)

````python
import numpy as np

#there is no probability for the image of being more 0 than 1

def clasify(x, P1_0, P1_1, P0 = 0.5, P1 = 0.5):
    x = np.array(x).flatten() #Matrix to unidimentional vector of 12 values
    l0 = np.prod(x*P1_0 + (1-x)*(1-P1_0))   # p(digit|0) #Product of all the probabilities of the pixels to see if we have 0
    l1 = np.prod(x*P1_1 + (1-x)*(1-P1_1))   # p(digito|1) #Product of all the probabilities of the pixels to see if we have 1
    with np.errstate(invalid='ignore', divide='ignore'): #posteriori probability, makes numpy ignore the 0/0 operations error
        post0 = (l0*P0) / (l0*P0 + l1*P1) # P(X|Y = 0) · P(Y = 0) / [P(X|Y = 0) · P(Y = 0) + P(X|Y = 1) · P(Y = 1)]
    return l0, l1, post0, 1-post0 # return products and probabilities
````

````python
new_digit = {
    "A": [[1,1,1,1],
                     [1,0,0,1],
                     [1,1,1,1]],
    "B": [[0,1,1,0],
                     [0,0,1,0],
                     [0,1,1,1]],
    "C":  [[1,1,0,1],
                     [0,1,0,0],
                     [1,1,1,0]],
}
````

````python
for name, d in new_digit.items():
    plt.imshow(np.array(d), cmap='binary'); plt.title(name); plt.show()
    l0, l1, p0, p1 = clasify(d, P1_0, P1_1)
    print(f"{name}: p(d|0)={l0:.6f}  p(d|1)={l1:.6f}  P(0|d)={p0:.4f}  P(1|d)={p1:.4f}")
````

````python
nondet  = (P1_0 == 0) & (P1_1 == 0)
alldet = (P1_0 == 1) & (P1_1 == 1)
x_nan = None
if nondet.any():
    x_nan = np.zeros(12)
    x_nan[nondet.argmax()] = 1
elif alldet.any():
    x_nan = np.ones(12)
    x_nan[alldet.argmax()] = 0

if x_nan is not None:
    l0, l1, p0, p1 = clasify(x_nan, P1_0, P1_1)
    print("p(d|0) =", l0, " p(d|1) =", l1, " P(0|d) =", p0)   # -> nan
else:
    print("not NaN.")

P1_0s = (digit0.sum(axis=0) + 1) / (len(digit0) + 2)
P1_1s = (digit1.sum(axis=0) + 1) / (len(digit1) + 2)
if x_nan is not None:
    print(" Laplace:", clasify(x_nan, P1_0s, P1_1s))
````

````python
for name, d in new_digit.items():
    _, _, p0_orig, _ = clasify(d, P1_0, P1_1)
    l0, l1, p0_s, p1_s = clasify(d, P1_0s, P1_1s)
    print(f"{name}: P(0|d)={p0_orig:.4f} | Laplace P(0|d)={p0_s:.4f}, P(1|d)={p1_s:.4f} -> prediction: {'0' if p0_s > p1_s else '1'}")
````

## Task 3 Conclusion

- Digit **A** is classified as **0**, and digit **B** is classified as **1**.
- Digit **C** produces **NaN**: pixel 5 is never "on" in any training example of 0 or 1, so p(C|0) = p(C|1) = 0 and the posterior becomes 0/0.
- With **Laplace smoothing** ((count + 1) / (total + 2)), the NaN disappears: P(0|C) = 0.64 and P(1|C) = 0.36, so C is classified as **0**, although with low confidence because the digit is ambiguous.
