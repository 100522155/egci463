```python
import pandas as pd

df1 = pd.read_csv("Iris.csv")
```

```python
print(df1.tail(10))
```

```python
width = df1.iloc[:, 1].tolist() #iloc = select numerical positions, in this case : -> all rows and 1 -> first column
print(width) # tolist = transform from matrix to list
```

```python
print(df1.head(10))
```

```python
 #Create a new column called Species_label
df1["Species_label"] = df1["Species"]
```

```python
df1["Species"] = df1["Species"].map({"Iris-setosa": 0, "Iris-versicolor": 1, "Iris-virginica": 2})
df1.head()
```

**TASK2**
![image.png]()
1.   Find mean/median/min/max/average for each column
2.   Print mean,var,min,max group by Species label
3.   Plot (scatter) each data based on each type



```python
media = df1.mean(numeric_only=True)
print(media)
median = df1.median(numeric_only=True)
print(median)
min = df1.min(numeric_only=True)
print(min)
max = df1.max(numeric_only=True)
print(max)
average = df1.mean(numeric_only=True)
print(average)
```

```python
result = df1.groupby("Species_label").agg(["mean", "var", "min", "max"])
#Groups made by the species label, agg means apply operations of "mean", "var", "min" & "max" inside the groups
print(result)
```

```python
import matplotlib.pyplot as plt

plt.scatter( #Representation in a graphic
    df1["SepalLengthCm"], # X edge with length
    df1["SepalWidthCm"], # Y edge wid width
    c=df1["Species_label"].map({"Iris-setosa": "Red", "Iris-versicolor": "Green", "Iris-virginica": "Yellow"})
    # Species labeled each as a color in the graphic
)
#Labels
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.title("Iris Data by Species")
plt.show() #Show the graphic
```
