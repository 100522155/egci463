# **Check point2:**
Choose (Weak) Classifier. Use only mean of 1 value to define the class

```python
y_pred=np.zeros(np.shape(y_test))
```

```python
y_pred
```

```python
def classifier(X_train,X_test):
  y_pred=np.zeros(np.shape(X_test[:,0]))
  y_pred[X_test[:,0]>np.mean(X_train[:,0])]=1
  y_pred[X_test[:,1]>np.mean(X_train[:,1])]=2
  return y_pred
```

```python
#New classifier function, only one mean value
def classifier2(X_train,X_test):
  y_pred=np.zeros(np.shape(X_test[:,0]))
  y_pred[X_test[:,0]>np.mean(X_train[:,0])]=1
  return y_pred
#This one with the other mean
def classifier3(X_train,X_test):
  y_pred=np.zeros(np.shape(X_test[:,0]))
  y_pred[X_test[:,1]>np.mean(X_train[:,1])]=2
  return y_pred
```

```python
Feature1=2
Feature2=3
y_pred=classifier(X_train[:,[Feature1,Feature2]],
                  X_test[:,[Feature1,Feature2]])
```

```python
print(y_pred)
```

```python
np.array([y_test, y_pred]).T
```

```python
accuracy = np.mean(y_pred == y_test)
print(f"Baseline accuracy: {accuracy:.2f}")
```

```python
#Now we verify how the classifier2 works
y_pred2=classifier2(X_train[:,[Feature1,Feature2]],
                  X_test[:,[Feature1,Feature2]])
```

```python
#Because we only have one mean value, we can only see the values that
#are better than the one mean. It it the minimun comparative point
print(y_pred2)
```

```python
#We can now see how much accuracy only one mean has
accuracy2 = np.mean(y_pred2 == y_test)
print(f"Baseline accuracy: {accuracy2:.2f}")
```

```python
#We can verify also  another classifier, with the other mean value
y_pred3=classifier3(X_train[:,[Feature1,Feature2]],
                  X_test[:,[Feature1,Feature2]])
```

```python
print(y_pred3)
```

```python
#Now we can see that both weak classifiers accuracy
accuracy3 = np.mean(y_pred3 == y_test)
print(f"Baseline accuracy: {accuracy2:.2f}")
```

```python
#Lastly we will try with a random number classifier
def classifier_random(X_train, X_test):
    n_samples = X_test.shape[0] #x_test length
    y_pred4 = np.random.choice([0, 1, 2], size=n_samples) #y_pred is an
    #x_test length array formed by 0,1 and 2. It is different each time
    #because its random
    return y_pred4
```

```python
y_pred_random = classifier_random(X_train[:,[Feature1,Feature2]],
                  X_test[:,[Feature1,Feature2]])
```

```python
print(y_pred_random)
#y_pred_random changes each time it is executed
```

```python
#the accuracy varies each time, it can be better than the other accuracies
#or worse
accuracy4 = np.mean(y_pred_random == y_test)
print(f"Baseline accuracy: {accuracy4:.2f}")
```
